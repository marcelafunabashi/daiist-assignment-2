# Train on Google Colab

Runs the same `train.py` as `uv run python main.py train`, but on a free GPU.

1. Runtime > Change runtime type > **T4 GPU**.
2. In the key icon on the left (Secrets), add `ROBOFLOW_API_KEY` and enable notebook access.
3. Set `REPO_URL` below to your fork, then Runtime > Run all.
4. When it finishes, `outputs.zip` downloads. Unzip it into the repo root (it holds `weights/` and `artifacts/`) and commit both.
